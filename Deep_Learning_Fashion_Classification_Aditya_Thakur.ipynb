# Deep Learning Practical: Fashion Image Classification

## Google Colab Practical for BBA AI/ML Students

### Business Scenario

Imagine that an e-commerce company receives thousands of product images every day. Instead of manually identifying every product, a Deep Learning model can classify images into categories such as T-shirt/Top, Trouser, Pullover, Dress, Coat, Sandal, Shirt, Sneaker, Bag, and Ankle Boot.

### Learning Objectives

- Understand how images can be used as input for Deep Learning.
- Build a simple Artificial Neural Network.
- Understand input, hidden and output layers.
- Train a model on product images.
- Check model accuracy.
- Use the trained model to predict a product category.
- Connect Deep Learning output with a business use case.


```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

print('TensorFlow version:', tf.__version__)
```

```python
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

print('Training images:', train_images.shape)
print('Testing images:', test_images.shape)
```

```python
class_names = [
    'T-shirt/Top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle Boot'
]
print(class_names)
```

```python
plt.imshow(train_images[0], cmap='gray')
plt.title('Product: ' + class_names[train_labels[0]])
plt.axis('off')
plt.show()
```

## Business Interpretation

Each image is a piece of data. For an e-commerce company:

**Input:** Product image  
**Expected Output:** Product category


```python
plt.figure(figsize=(10, 6))
for i in range(12):
    plt.subplot(3, 4, i + 1)
    plt.imshow(train_images[i], cmap='gray')
    plt.title(class_names[train_labels[i]])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
train_images = train_images / 255.0
test_images = test_images / 255.0
print('Images are ready for training.')
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
print('Model is ready for training.')
```

```python
history = model.fit(
    train_images,
    train_labels,
    epochs=3,
    validation_split=0.1
)
```

```python
test_loss, test_accuracy = model.evaluate(test_images, test_labels, verbose=0)
print('Test Accuracy:', round(test_accuracy * 100, 2), '%')
```

```python
predictions = model.predict(test_images, verbose=0)
predicted_class = np.argmax(predictions[0])
actual_class = test_labels[0]
print('Predicted Product:', class_names[predicted_class])
print('Actual Product:', class_names[actual_class])
```

```python
plt.imshow(test_images[0], cmap='gray')
plt.title('Predicted: ' + class_names[predicted_class] + '\nActual: ' + class_names[actual_class])
plt.axis('off')
plt.show()
```

```python
image_number = 25
prediction = model.predict(np.expand_dims(test_images[image_number], axis=0), verbose=0)
predicted_class = np.argmax(prediction)
actual_class = test_labels[image_number]
plt.imshow(test_images[image_number], cmap='gray')
plt.title('Predicted: ' + class_names[predicted_class] + '\nActual: ' + class_names[actual_class])
plt.axis('off')
plt.show()
```
